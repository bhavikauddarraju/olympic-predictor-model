```python
import zipfile
import pandas as pd

# Unzip the file automatically in Colab
with zipfile.ZipFile('archive (1).zip', 'r') as zip_ref:
    zip_ref.extractall('.')

print("Extracted successfully!")

# Load the dataset
df = pd.read_csv('athlete_events.csv')

# Display the first few rows
display(df.head())
```

```python
# Filter for only Summer Olympics to keep things consistent
summer_df = df[df['Season'] == 'Summer']

# Group by Country (NOC) and Year to calculate total athletes and total medals won
# We create a column 'Won_Medal' which is 1 if they won a medal, 0 otherwise
summer_df['Won_Medal'] = summer_df['Medal'].notna().astype(int)

team_performance = summer_df.groupby(['NOC', 'Year']).agg(
    total_athletes=('ID', 'nunique'),
    total_medals=('Won_Medal', 'sum')
).reset_index()

# View the first few rows of our new aggregated table
print("Aggregated country-level data:")
display(team_performance.head())
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

# Drop rows where previous medals might be missing (just to be safe)
ml_data = team_performance.dropna()

# 1. Split data by Year to avoid data leakage (train on past games, test on recent games)
# Let's train on Olympics up to 2012, and test on 2016
train = ml_data[ml_data['Year'] < 2016]
test = ml_data[ml_data['Year'] >= 2016]

# 2. Define our Features (Inputs) and Target (Output)
features = ['total_athletes', 'prev_medals']
target = 'total_medals'

X_train = train[features]
y_train = train[target]

X_test = test[features]
y_test = test[target]

# 3. Initialize and train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# 4. Make predictions on the test set
predictions = model.predict(X_test)

# Add predictions to our test set to look at them
test_results = X_test.copy()
test_results['actual_medals'] = y_test
test_results['predicted_medals'] = predictions

# 5. Evaluate how well the model did using Mean Absolute Error (MAE)
mae = mean_absolute_error(y_test, predictions)
print(f"Model Evaluation - Mean Absolute Error: {mae:.2f} medals")
print("\nHere are a few predictions compared to reality:")
display(test_results.head(10))
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

# 1. Filter for Summer Olympics cleanly
summer_df = df[df['Season'] == 'Summer'].copy()

# 2. Create the medal indicator column
summer_df['Won_Medal'] = summer_df['Medal'].notna().astype(int)

# 3. Group by Country and Year
team_performance = summer_df.groupby(['NOC', 'Year']).agg(
    total_athletes=('ID', 'nunique'),
    total_medals=('Won_Medal', 'sum')
).reset_index()

# 4. Create the past medals lag feature
team_performance['prev_medals'] = team_performance.groupby('NOC')['total_medals'].shift(1)
team_performance['prev_medals'] = team_performance['prev_medals'].fillna(0)

# Drop missing rows
ml_data = team_performance.dropna()

# 5. Split data into Training (< 2016) and Testing (>= 2016)
train = ml_data[ml_data['Year'] < 2016]
test = ml_data[ml_data['Year'] >= 2016]

# 6. Define Features and Target
features = ['total_athletes', 'prev_medals']
target = 'total_medals'

X_train, y_train = train[features], train[target]
X_test, y_test = test[features], test[target]

# 7. Train Linear Regression Model
model = LinearRegression()
model.fit(X_train, y_train)

# 8. Predict and Evaluate
predictions = model.predict(X_test)

test_results = X_test.copy()
test_results['actual_medals'] = y_test
test_results['predicted_medals'] = predictions

mae = mean_absolute_error(y_test, predictions)
print(f"🎉 Success! Model Evaluation - Mean Absolute Error: {mae:.2f} medals")
print("\nHere are a few predictions compared to reality:")
display(test_results.head(10))
```

```python
# Clean up the predictions: fix negative numbers and show country codes
cleaned_results = test.copy()
cleaned_results['predicted_medals'] = model.predict(X_test)

# Fix negative predictions (set anything below 0 to 0) and round to whole numbers
cleaned_results['predicted_medals'] = cleaned_results['predicted_medals'].clip(lower=0).round().astype(int)

# Display country, year, past medals, actual medals, and what our model predicted
display(cleaned_results[['NOC', 'Year', 'prev_medals', 'total_medals', 'predicted_medals']].head(10))
```
